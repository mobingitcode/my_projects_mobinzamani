```python
# فراخوانی کتابخانه
import cv2
import matplotlib.pyplot as plt
# تعریف کلاس 
class picture() :     
    # تعریف تابع اولیه و سازنده 
    def __init__(self , path) : 
        self.path = path 
        self.img = cv2.imread(self.path)
        self.img_rgb = cv2.cvtColor(self.img, cv2.COLOR_BGR2RGB)
        self.img_gray = cv2.cvtColor(self.img, cv2.COLOR_BGR2GRAY)
        self.fig = plt.figure(figsize=(15,15))
   
    # اجرای سطر اول (خود تصویر و تصویر قرمز)
    def display_original_and_red_channel(self) : 
        # نمایش خود تصویر 
        plt.subplot(4 , 2 , 1)
        plt.imshow(self.img_rgb)
        plt.title("Original Image")
        plt.axis('off')
        # استخراچ کانال قرمز و دو برابر کردن مقدار پیکسل ها (در صورتی که از 255 بیشتر شد همان 255)
        red_channel = self.img_rgb[:, :, 0]
        double = cv2.multiply(red_channel , 2)
        double[double > 255] = 255
        self.img_rgb[:, :, 0] = double
        plt.subplot(4 , 2 , 2)
        plt.imshow(self.img_rgb)  
        plt.title("img_red")
        plt.axis('off')
    # اجرای سطر دوم (خاکستری تصویر قرمز و هیستوگرام آن) 
    def display_gray_and_histogram(self) : 
        double_gray = cv2.cvtColor(self.img_rgb, cv2.COLOR_RGB2GRAY) # خاکستری کردن 
        plt.subplot(4 , 2 , 3)
        plt.imshow(double_gray, cmap='gray')
        plt.title("Gray of IMG-RED")
        plt.axis('off')
        plt.subplot(4 , 2 , 4) # هیستوگرام 
        plt.hist(double_gray.ravel(), bins=256, range=(0, 256), color='red')
        plt.title("Histogram_gray")

    def display_swap_black_white(self) : 
        # جابجایی رنگ های سفید و سیاه
        swapped_image = self.img_rgb.copy()
        black_pixels = (swapped_image[:,:,0] == 0) & (swapped_image[:,:,1] == 0) & (swapped_image[:,:,2] == 0)
        white_pixels = (swapped_image[:,:,0] == 255) & (swapped_image[:,:,1] == 255) & (swapped_image[:,:,2] == 255)
        swapped_image[black_pixels] = [255, 255, 255]
        swapped_image[white_pixels] = [0, 0, 0]
        
        swapped_image_gray = cv2.cvtColor(swapped_image, cv2.COLOR_RGB2GRAY)
       
        # رسم تصویر 
        plt.subplot(4 , 2 , 5)
        plt.imshow(swapped_image_gray , cmap="gray")
        plt.title("Black<->White Swapped")
        plt.axis('off')
        # رسم هیستوگرام
        plt.subplot(4 , 2 , 6)
        plt.hist(swapped_image_gray.ravel(), bins=256, range=(0, 256), color='blue')
        plt.title("Histogram_swap")
   
    # اجرای سطر 4 (معکوس کردن شدت روشنایی عکس سطر دو و رسم هیستوگرام)
    def display_inverse_intensity(self) :
        
        inverted_image = 255 - self.img_rgb
        inverted_image_gray = cv2.cvtColor(inverted_image, cv2.COLOR_RGB2GRAY)
        plt.subplot(4 , 2 , 7)
        plt.imshow(inverted_image_gray,cmap="gray") # رسم عکس
        plt.title("inverse")
        plt.axis('off')
        
        plt.subplot(4 , 2 , 8)
        plt.hist(inverted_image_gray.ravel(), bins=256, range=(0, 256), color='green') # رسم هیستوگرام 
        plt.title("Histogram_inverse")

my_pic = picture(r"C:\Users\PISHRORAYANEH\Desktop\project 5\landscapeimg.jpg")

my_pic.display_original_and_red_channel()
my_pic.display_gray_and_histogram()
my_pic.display_swap_black_white()
my_pic.display_inverse_intensity()

plt.show()
```
