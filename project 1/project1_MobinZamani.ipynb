```python
received_msg = "recived_msg.txt"
sent_msg = "sent_msg.txt"
# فراخوانی و خواندن فایل ها
with open(received_msg, "r", encoding = "utf-8") as file:
    received_message = file.read()
# پیدا کردن اسم در فایل 
received_message_list = received_message.splitlines()
user_name = received_message_list[2]

with open(sent_msg, "r", encoding = "utf-8") as file:
    sent_message = file.read()
#قرار دادن اسم کاربر در فایل ارسالی 
sent_message_list = sent_message.split()

sent_message_list[1] = user_name + ','
# تنظیم متن ارسالی از لیست
new_text = ' '.join(sent_message_list )
print(new_text)



```
