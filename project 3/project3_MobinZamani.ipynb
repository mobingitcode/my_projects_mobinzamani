```python
import numpy as np
import matplotlib.pyplot as plt
def Blue_sine(x, file_path):
        y = np.sin(x)
        plt.figure(figsize=(8, 6))
        plt.plot(x, y, color='blue', linewidth=2, label='y = sin(x)')
        plt.title('Sine Wave Plot')
        plt.xlabel('x')
        plt.ylabel('sin(x)')
        plt.legend()
        plt.grid(True)
        plt.savefig(file_path , format = "PNG")
        plt.close()
x = np.linspace(-1, 1, 100)
file_path = input()

Blue_sine(x, file_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
def Red_sine(x, file_path):
        y = np.sin(x)
        plt.figure(figsize=(8, 6))
        plt.plot(x, y, color='Red', linewidth=2, label='y = sin(x)')
        plt.title('Sine Wave Plot')
        plt.xlabel('x')
        plt.ylabel('sin(x)')
        plt.legend()
        plt.grid(True)
        plt.savefig(file_path , format = "PNG")
        plt.close()
x = np.linspace(-1, 1, 100)
file_path = input()

Red_sine(x, file_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def Both_sine(x, file_path):
    y1 = np.sin(x)
    y2 = -np.sin(x)

    plt.figure(figsize=(8, 6))
    plt.plot(x, y1, color='blue', linewidth=2, label='y = sin(x)')
    plt.plot(x, y2, color='red', linewidth=2, label='y = -sin(x)')

    plt.title('Sine and Negative Sine Waves')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)
    plt.savefig(file_path, format="PNG")
    plt.close()

x = np.linspace(-1, 1, 100)
file_path = input()

Both_sine(x, file_path)
```
